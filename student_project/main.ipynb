```python
from students import generate_students
from analysis import add_ages, calculate_statistics
from pathlib import Path
```

```python
students = generate_students(10)
students = add_ages(students)

for student in students:
    print(student)
```

```python
stats = calculate_statistics(students)

print("CLASS STATISTICAL ANALYSIS")
print("-" * 30)

for name, value in stats.items():
    print(f"{name}: {value}")
```

```python

from pathlib import Path

project_folder = Path.cwd() / "student_project"
results_folder = project_folder / "results"
results_folder.mkdir(parents=True, exist_ok=True)

report_path = results_folder / "report.txt"

with report_path.open("w", encoding="utf-8") as file:
    file.write("STUDENT CLASS ANALYSIS REPORT\n")
    file.write("=" * 35 + "\n\n")

    for name, value in stats.items():
        file.write(f"{name}: {value}\n")

    file.write("\nSTUDENT DETAILS\n")
    for student in students:
        file.write(f"{student}\n")

print("Report saved successfully:", report_path)
```
